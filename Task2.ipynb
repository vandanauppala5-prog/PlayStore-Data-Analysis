```python
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz
```

```python
# Load Playstore dataset
apps = pd.read_csv(r"C:\Users\vanda\Downloads\Play Store Data.csv")
apps.head()     
```

```python
# Data Cleaning
apps["Installs"] =(apps["Installs"].astype(str).str.replace(",","",regex = False).str.replace("+","",regex = False))
```

```python
apps["Installs"] = pd.to_numeric(apps["Installs"],errors = "coerce")
```

```python
apps = apps.dropna(subset = ["Installs"])
```

```python
apps["Installs"] = apps["Installs"].astype(int)
```

```python
# Filter Data
apps = apps[~apps["Category"].str.upper().str.startswith(("A","C","G","S"),na=False)]
```

```python
apps = apps[apps["Installs"] > 1000000]
```

```python
# TOP 5
top5 = (apps.groupby("Category")["Installs"].sum().nlargest(5).reset_index())
```

```python
map_apps = apps.merge(top5[["Category"]],on = "Category")
```
